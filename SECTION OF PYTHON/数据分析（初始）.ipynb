```python
# import pandas as pd
# import matplotlib.pyplot as plt
# from IPython.display import display
# pd.set_option('display.unicode.east_asian_width',True)
# pd.set_option('display.unicode.ambiguous_as_wide',True)
# pd.set_option('display.max_column',None)
# pd.set_option('display.width',1500)
# pd.set_option('display.expand_frame_repr',False)
# pd.set_option('display.max_colwidth',30)
# df=pd.read_csv('C:\\Users\\34979\\Desktop\\stats_tables_clean.csv')
# df
# df2=df.head(17)
# df2
# df2.columns=df2.iloc[0]
# df2=df2.drop(0)
# df2=df2.reset_index(drop=True)
# df2
# df2 = df2[['行业', '绝对额-二季度(亿元)', '绝对额-上半年(亿元)', '同比增长-二季度(%)', '同比增长-上半年(%)']]
# numeric_cols = ['绝对额-二季度(亿元)', '绝对额-上半年(亿元)', '同比增长-二季度(%)', '同比增长-上半年(%)']
# for col in numeric_cols:
#     df2[col] = pd.to_numeric(df2[col], errors='coerce')
# df_sorted = df2.sort_values(
#     by=['同比增长-二季度(%)'],  # 按所有数值列排序
#     ascending=[False] 
# )
# display(df_sorted)
# y=[i for i in df_sorted['同比增长-二季度(%)']]
# x=[j for j in df_sorted['行业']]
# plt.plot(x,y)
# plt.show()
# df=df.sort_values(['绝对额-二季度(亿元)','绝对额-上半年(亿元)','同比增长-二季度(%)','同比增长-上半年(%)'],ascending=[False,False,False,False])
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# 设置pandas显示选项
pd.set_option('display.unicode.east_asian_width', True)
pd.set_option('display.unicode.ambiguous_as_wide', True)
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1500)
pd.set_option('display.expand_frame_repr', False)
pd.set_option('display.max_colwidth', 30)

# 设置matplotlib中文字体（解决乱码问题）
plt.rcParams['font.sans-serif'] = ['SimHei', 'Microsoft YaHei', 'Arial Unicode MS']  # 设置中文字体
plt.rcParams['axes.unicode_minus'] = False  # 解决负号显示问题

# 读取数据
df = pd.read_csv('C:\\Users\\34979\\Desktop\\stats_tables_clean.csv')

# 提取前17行（表1的数据）
df2 = df.head(17)

# 设置列名
df2.columns = df2.iloc[0]
df2 = df2.drop(0)
df2 = df2.reset_index(drop=True)

# 筛选需要的列
df2 = df2[['行业', '绝对额-二季度(亿元)', '绝对额-上半年(亿元)', '同比增长-二季度(%)', '同比增长-上半年(%)']]

# 转换数值列
numeric_cols = ['绝对额-二季度(亿元)', '绝对额-上半年(亿元)', '同比增长-二季度(%)', '同比增长-上半年(%)']
for col in numeric_cols:
    df2[col] = pd.to_numeric(df2[col], errors='coerce')

# 按二季度同比增长率降序排序
df_sorted = df2.sort_values(
    by=['同比增长-二季度(%)'],
    ascending=[False]
)

# 显示排序后的数据
display(df_sorted)

# ===== 画图部分（使用排序后的数据）=====
x = df_sorted['行业'].tolist()  # 行业名称
y = df_sorted['同比增长-二季度(%)'].tolist()  # 二季度同比增长率

# 创建图表
plt.figure(figsize=(12, 6))
plt.plot(x, y, marker='o', linestyle='-', linewidth=2, markersize=8)

# 添加标题和标签
plt.title('各行业二季度同比增长率（降序排列）', fontsize=16)
plt.xlabel('行业', fontsize=12)
plt.ylabel('同比增长率（%）', fontsize=12)

# 旋转x轴标签，避免重叠
plt.xticks(rotation=45, ha='right', fontsize=10)

# 添加网格线
plt.grid(True, alpha=0.3, linestyle='--')

# 显示数值标签
for i, v in enumerate(y):
    plt.text(i, v + 0.2, f'{v:.1f}%', ha='center', va='bottom', fontsize=8)

# 调整布局，防止标签被裁剪
plt.tight_layout()

# 显示图表
plt.show()
```
