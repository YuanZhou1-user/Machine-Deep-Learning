```python
import numpy as np
from sklearn.linear_model import LinearRegression
x=np.array([[1.5],[1.6],[1.7]])
y=np.array([90,100,110])
model=LinearRegression()
model.fit(x,y)
new_height=[[1.65]]
predicted_weight=model.predict(new_height)
print(predicted_weight[0])
```

```python
from sklearn.preprocessing import StandardScaler
import numpy as np
x_train=np.array([[25,8000],
                  [30,12000],
                  [35,5000]])
x_test=[[28,10000]]
#创建工具
scaler=StandardScaler()
scaler.fit(x_train)
print(f"{scaler.mean_[0]:.4f}")
print(f"{scaler.scale_[0]:.4f}")
#转换(特征缩放)
x_train_scaled=scaler.transform(x_train)
x_test_scaled=scaler.transform(x_test)
print(x_train_scaled)
print(x_test_scaled)
```

```python
#SGDRegressor :(随机梯度下降)
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import SGDRegressor
from sklearn.model_selection import train_test_split
X = [[50, 3], [80, 5], [100, 8], [60, 2], [120, 10], [70, 4]]
y = [150, 240, 310, 180, 380, 210]
x_train,x_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=42)
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
model=SGDRegressor(max_iter=1000,tol=1e-3,learning_rate='adaptive',eta0=0.01,random_state=42)
model.fit(x_train_scaled,y_train)
score=model.score(x_train_scaled,y_train)
print(f"{score:.4f}")
```
