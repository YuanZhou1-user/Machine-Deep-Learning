```python
import numpy as np
#引入模型
from sklearn.linear_model import LogisticRegression
X = np.array([[0.5, 1.5], [1,1], [1.5, 0.5], [3, 0.5], [2, 2], [1, 2.5]])
y = np.array([0, 0, 0, 1, 1, 1])

lr_model=LogisticRegression()

#利用x,y训练模型
lr_model.fit(X,y)

#模型预测值
lr_pred=lr_model.predict([[1.4,0.45]])
lr_pred1=lr_model.predict(X)

#模型准确率
lr_accu=lr_model.score(X,y)
lr_accu1=lr_model.score([[1.4,0.45]],[0])
print(lr_model)
print(lr_pred)
print(lr_pred1)
print(lr_accu)
print(lr_accu1)
```
